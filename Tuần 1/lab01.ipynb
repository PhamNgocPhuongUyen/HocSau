```python
# Bài 6. Phép toán và nhân ma trận
##Thực hiện các phép toán phần tử và phép nhân ma trận bằng toán tử @.
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na + b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
print("Shape của a @ b.T:", (a @ b.T).shape)
```

Phân biệt phép toán a * b với a @ b.T

a * b : Nhân từng phần tử tương ứng của hai Tensor có cùng kích thước. Kết quả ra Tensor có kích thước giống a và b (2x3)

a @ b.T: Phép nhân ma trận đại số tuyến tính giữa ma trận a kích thước (2x3)  và ma trận chuyển vị $b^T$ kích thước (3x2)

```python
#Bài 7. Các hàm thống kê trên Tensor
##Tính trung bình, độ lệch chuẩn và tổng tích lũy theo từng cột.
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))

print("Mean (dim=1):")
print(torch.mean(a, dim=1))
print("\nStandard deviation (dim=1):")
print(torch.std(a, dim=1))
print("\nCumulative sum (dim=1):")
print(torch.cumsum(a, dim=1))
```

Khi dim=0 (Tính theo Cột):

+Phép toán duyệt dọc theo các hàng để tính cho từng cột

Khi dim=1 (Tính theo Hàng):

+cumsum cộng dồn từ trái sang phải theo từng hàng

```python
#Bài 8. Autograd cơ bản
##Tạo biến có requires_grad=True và dùng backward() để tính đạo hàm tự động.
import torch
x = torch.tensor(
3.6,
requires_grad=True
)
y = x * x
y.backward()
print("x =", x)
print("y =", y)
print("x.grad =", x.grad)

# Khai báo lại biến x (hoặc xóa gradient cũ bằng x.grad.zero_())
x = torch.tensor(3.6, requires_grad=True)
# Định nghĩa hàm mới y = 3*x**2 + 2*x + 1
y = 3 * x**2 + 2 * x + 1
# Tính gradient
y.backward()
print("y =", y)
print("x.grad =", x.grad)
```

```python
#Bài 9. Gradient Descent tìm hệ số đa thức
##Sinh dữ liệu từ đa thức y = x2 + 2x + 3, sau đó dùng NAdam và autograd để tìm lại ba hệ số.
import numpy as np
import torch
polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
X * X,
X,
np.ones_like(X)
])
w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)
print("Initial coefficients:")
print(w)
for epoch in range(1000):
 optimizer.zero_grad()
 y_pred = x @ w
 mse = torch.mean(torch.square(y - y_pred))
 mse.backward()
 optimizer.step()
print("Learned coefficients:")
print(w)
```

**So sánh:**

Hệ số $a$ (của $x^2$) $\approx 1.0177$ xấp xỉ đúng với $1$.

Hệ số $b$ (của $x$) $\approx 2.0003$ xấp xỉ chính xác với $2$.

Hệ số $c$ (hằng số) $\approx 2.1395$ chênh lệch nhẹ so với $3$ do số lượng epoch (1000) và hằng số học ($lr=0.01$) chưa đủ để tham số hằng số hội tụ hoàn toàn.

**Giải thích vai trò:**

optimizer.zero_grad(): Xóa (reset) toàn bộ gradient tích lũy của các tham số về $0$ trước khi tính toán vòng lặp mới, tránh việc gradient của các epoch trước bị cộng dồn vào epoch hiện tại.

mse.backward(): Sử dụng thuật toán Autograd để tính đạo hàm (gradient) của hàm mất mát mse theo trọng số w ($\frac{\partial \text{loss}}{\partial w}$) thông qua Lan truyền ngược 

optimizer.step(): Cập nhật lại giá trị trọng số w dựa trên gradient vừa tính được và thuật toán tối ưu NAdam ($w_{\text{mới}} = w_{\text{cũ}} - \Delta w$)

```python
#Bài 10. Giải hệ phương trình bằng Autograd
##Tối thiểu hóa tổng sai số bình phương để tìm A, B, C, D thỏa bốn phương trình.
import random
import torch
A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)
for _ in range(EPOCHS):
 y1 = A + B - 9
 y2 = C - D - 1
 y3 = A + C - 8
 y4 = B - D - 2
 sqerr = y1*y1 + y2*y2 + y3*y3 + y4*y4
 optimizer.zero_grad()
 sqerr.backward()
 optimizer.step()
print("A =", A)
print("B =", B)
print("C =", C)
print("D =", D)
```

Thay các giá trị thu được từ chương trình ($A \approx 4.6255, B \approx 4.3744, C \approx 3.3745, D \approx 2.3745$) vào hệ:

-$A + B = 9$: $4.6255 + 4.3744 = 8.9999 \approx 9$ (Thỏa mãn)

-$C - D = 1$: $3.3745 - 2.3745 = 1.0000 = 1$ (Thỏa mãn)

-$A + C = 8$: $4.6255 + 3.3745 = 8.0000 = 8$ (Thỏa mãn)

-$B - D = 2$: $4.3744 - 2.3745 = 1.9999 \approx 2$ (Thỏa mãn)

**Nhận xét vì sao bài toán có thể có nhiều nghiệm**

-Tính phụ thuộc tuyến tính: Nếu lấy phương trình (1) trừ phương trình (3) ta được $B - C = 1$. Lấy phương trình (2) cộng phương trình (4) ta cũng được $B - C = 1$. Hai vế này trùng nhau, dẫn đến hệ 4 phương trình nhưng thực chất chỉ có 3 phương trình độc lập.

-Hệ vô số nghiệm: Do số ẩn ($4$) nhiều hơn số phương trình độc lập ($3$), hệ phương trình có vô số nghiệm phụ thuộc vào một tham số tự do (ví dụ biểu diễn theo $D$: $C = D + 1$, $B = D + 2$, $A = 7 - D$).

-Mỗi lần chạy lại chương trình, random.random() sẽ khởi tạo điểm bắt đầu khác nhau cho $A, B, C, D$. Tùy vào giá trị ngẫu nhiên ban đầu này, thuật toán NAdam sẽ hội tụ về một nghiệm cụ thể khác nhau trong tập vô số nghiệm của hệ.

```python
## Cho y = 2x2 + 5x + 3. Dùng autograd tính dy/dx tại x = 2 và kiểm tra bằng phép tính thủ công.

import torch
# Khai báo biến x = 2 với requires_grad=True
x = torch.tensor(2.0, requires_grad=True)
y = 2 * x**2 + 5 * x + 3
# Tính đạo hàm tự động bằng autograd
y.backward()
# In kết quả từ PyTorch
autograd_res = x.grad.item()
print(f"Giá trị dy/dx (PyTorch Autograd) tại x = 2: {autograd_res}")
# Kiểm tra bằng phép tính thủ công
# y' = d/dx (2x^2 + 5x + 3) = 4x + 5
# Tại x = 2: y'(2) = 4*(2) + 5 = 13
manual_res = 4 * 2.0 + 5
print(f"Giá trị dy/dx (Thủ công) tại x = 2: {manual_res}")
print("Kết quả trùng khớp:", autograd_res == manual_res)
```

```python
## Thay đa thức trong Bài 9 bằng y = 2x2 - 3x + 5; huấn luyện và cho biết các hệ số tìm được.
import numpy as np
import torch

# Thay đa thức thành y = 2x^2 - 3x + 5
polynomial = np.poly1d([2, -3, 5])
N = 20

X = np.random.randn(N, 1) * 5
Y = polynomial(X)

XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)

optimizer = torch.optim.NAdam([w], lr=0.01)

print("Initial coefficients:")
print(w)

# Huấn luyện 2000 epoch để các hệ số hội tụ tốt hơn
for epoch in range(3000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("\nLearned coefficients:")
print(w)
```
